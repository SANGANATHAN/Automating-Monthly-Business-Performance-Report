```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace(r"\s+","_",regex=True)
    .str.replace(r"[^A-Za-z0-9_]","",regex=True)
    .str.lower()
)
print("Column Names Standardized Successfully!")
print("\nUpdated Column Names:")
print(df.columns.tolist())
```

```python
duplicate_records = df[df.duplicated(keep=False)]
print("Total Duplicate Records:",len(duplicate_records))
```

```python
before_count = len(df)
df = df.drop_duplicates(keep="first")
after_count = len(df)
removed_count = before_count - after_count
print("Duplicate Removal Summary")
print("--------------------------")
print("Before Cleaning:",before_count)
print("After Cleaning:",after_count)
print("Total Rows Removed:",removed_count)
```

```python
duplicate_ids = df[
    df.duplicated(subset=["transaction_id"],keep=False)
    ]
print("Duplicate Transaction ID Records:",len(duplicate_ids))
display(duplicate_ids.head(10))
```

```python
unwanted_columns = ["temp_code", "imported_by"]
df = df.drop(columns=unwanted_columns)
print("Unwanted Columns Removed Successfully!")
print("\nUpdated Dataset Shape:",df.shape)
print("\nRemaining Columns:")
print(df.columns.tolist())
```

```python
text_columns = df.select_dtypes(include=["str","object"]).columns
for column in text_columns:
    df[column] = (
        df[column]
        .astype("string")
        .str.strip()
        .str.replace(r"\s+"," ",regex=True)
        .str.replace(r"[^a-zA-Z]", "", regex=True)
    )
print("Extra Spaces Removed Successfully!")
```

```python
customer_split = df["customer_details"].str.split(
    "|",n=1,expand=True,regex=False
)
df["customer_id"] = customer_split[0].str.strip()
df["customer_type"] = customer_split[1].str.strip()
print("Customer Details Split Successfully!")
display(df[["customer_id","customer_type"]].head(10))

product_split = df["product_details"].str.split(
    "|",n=2,expand=True,regex=False
)
df["product_id"] = product_split[0].str.strip()
df["product_name"] = product_split[1].str.strip()
df["product_category"] = product_split[2].str.strip()
print("Product Details Split Successfully!")
display(df[[
    "product_id",
    "product_name",
    "product_category"
]].head(10))
```

```python
df = df.drop(columns=[
    
    "internal_remarks"
])
print("Old Combined Columns Removed!")
print("Updated Dataset Shape:",df.shape)
print("\nUpdated Columns:")
print(df.columns.tolist())
```

```python
check_columns =[
    "customer_type",
    "product_category",
    "order_status",
    "region",
    "sales_channel"
]
for column in check_columns:
    print("\n",column.upper())
    print(df[column].value_counts(dropna=False))
```

```python
status_mapping = {
    "completed": "Completed",
    "@@completed##": "Completed",
    "@@ completed ##": "Completed",
    "cancelled": "Cancelled",
    "canceld": "Cancelled",
    "cance lled": "Cancelled",
    "pending": "Pending",
    "pendng": "Pending",
    "@@returned##": "Returned",
    "returned": "Returned"
}
region_mapping = {
    "tirunelveli": "Tirunelveli",
    "madurai": "Madurai",
    "maduraii": "Madurai",
    "chen nai": "Chennai",
    "chennai": "Chennai",
    "coimbatore": "Coimbatore",
    "coim bator": "Coimbatore",
    "salem": "Salem",
    "saalem": "Salem",
    "trichy": "Trichy"
}
channel_mapping = {
    "online": "Online",
    "@@online##": "Online",
    "store": "Store",
    "@@store##": "Store",
    "mobile app": "Mobile App",
    "@@mobile app##": "Mobile App"
}
df["order_status"] = (
    df["order_status"].str.strip().str.lower()
    .map(status_mapping).fillna(df["order_status"])
)

df["region"] = (
    df["region"].str.strip().str.lower()
    .map(region_mapping).fillna(df["region"])
)

df["sales_channel"] = (
    df["sales_channel"].str.strip().str.lower()
    .map(channel_mapping).fillna(df["sales_channel"])
)

print("Text Standardization Completed!")
```

```python
numeric_columns = [
    "quantity",
    "unit_price",
    "discount",
    "revenue",
    "expenses",
    "profit"
]
for column in numeric_columns:
    df[column] = (
        df[column]
        .astype("string")
        .str.replace(r"[₹,$,]","",regex=True)
        .str.strip()
    )
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )
print("Numeric Conversion Completed!")
print("\nUpdated Data Types:")
print(df[numeric_columns].dtypes)
print("\nNumerice Data Preview:")
display(df[numeric_columns].head(10))
```

```python
print("Date Column Data Type:")
print(df["date"].dtype)
print("\nSample Date Values:")
display(df["date"].dropna().head(20))
print("\nUnique Date Formats:")
print(df["date"].dropna().astype(str).str.len().value_counts())
```

```python
date_text = df["date"].astype("string").str.strip()
converted_dates = pd.Series(
    pd.NaT,
    index=df.index,
    dtype="datetime64[ns]"
)
iso_mask = date_text.str.match(
    r"^\d{4}-\d{2}-\d{2}$",
    na=False
)
converted_dates.loc[iso_mask] = pd.to_datetime(
    date_text.loc[iso_mask],
    format="%Y-%m-%d",
    errors="coerce"
)
numeric_mask = date_text.str.match(
    r"^\d{2}-\d{2}-\d{4}$",
    na=False
)
converted_dates.loc[numeric_mask] = pd.to_datetime(
    date_text.loc[numeric_mask],
    format="%m-%d-%Y",
    errors="coerce"
)

text_mask = date_text.str.match(
    r"^\d{2}-[A-Za-z]{3}-\d{4}$",
    na=False
)

converted_dates.loc[text_mask] = pd.to_datetime(
    date_text.loc[text_mask],
    format="%d-%b-%Y",
    errors="coerce"
)
df["date"] = converted_dates
print("Date Conversion Completed!")
print("\nUpdated Date Data Type:")
print(df["date"].dtype)
print("\nConverted Date Preview:")
display(df[["date"]].head(20))
```

```python

df.dropna(inplace=True)
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nFirst Five Rows:")
display(df.head())
```
