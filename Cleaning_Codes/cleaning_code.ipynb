```python

# STEP 16.2: Standardize Text Values

# Order Status Mapping
status_mapping = {
    "completed": "Completed",
    "@@completed##": "Completed",
    "@@ completed ##": "Completed",
    "cancelled": "Cancelled",
    "canceld": "Cancelled",
    "cance lled": "Cancelled",
    "pending": "Pending",
    "pendng": "Pending",
    "@@returned##": "Returned",
    "returned": "Returned"
}

# Region Mapping
region_mapping = {
    "tirunelveli": "Tirunelveli",
    "madurai": "Madurai",
    "maduraii": "Madurai",
    "chen nai": "Chennai",
    "chennai": "Chennai",
    "coimbatore": "Coimbatore",
    "coim bator": "Coimbatore",
    "salem": "Salem",
    "saalem": "Salem",
    "trichy": "Trichy"
}

# Sales Channel Mapping
channel_mapping = {
    "online": "Online",
    "@@online##": "Online",
    "store": "Store",
    "@@store##": "Store",
    "mobile app": "Mobile App",
    "@@mobile app##": "Mobile App"
}

# Apply mappings
df["order_status"] = (
    df["order_status"].str.strip().str.lower()
    .map(status_mapping).fillna(df["order_status"])
)

df["region"] = (
    df["region"].str.strip().str.lower()
    .map(region_mapping).fillna(df["region"])
)

df["sales_channel"] = (
    df["sales_channel"].str.strip().str.lower()
    .map(channel_mapping).fillna(df["sales_channel"])
)

print("Text Standardization Completed!")
```

```python

numeric_columns = [
    "quantity",
    "unit_price",
    "discount",
    "revenue",
    "expenses",
    "profit"
]

outlier_results = []

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_results.append({
        "Column": column,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": len(outliers)
    })

outlier_report = pd.DataFrame(outlier_results)

print("IQR Outlier Detection Report:")
display(outlier_report)
```

```python

import numpy as np

df["data_quality_issue"] = ""

conditions = {
    "Negative Quantity": df["quantity"] < 0,
    "Zero Unit Price": df["unit_price"] == 0,
    "Negative Unit Price": df["unit_price"] < 0,
    "Negative Expenses": df["expenses"] < 0,
    "Suspicious Expenses": df["expenses"] >= 99999999,
    "High Quantity": df["quantity"] > 100
}

for issue, condition in conditions.items():
    df.loc[condition, "data_quality_issue"] = (
        df.loc[condition, "data_quality_issue"]
        .apply(lambda x: x + "; " if x else "") + issue
    )

issue_report = df[df["data_quality_issue"] != ""]

print("Total Records with Numeric Issues:", len(issue_report))

display(issue_report[
    ["transaction_id", "quantity", "unit_price",
     "revenue", "expenses", "profit", "data_quality_issue"]
])
```

```python

# Preserve original values for audit
df["original_quantity"] = df["quantity"]
df["original_unit_price"] = df["unit_price"]
df["original_expenses"] = df["expenses"]

# 1. Correct invalid unit prices using Product Master
invalid_price = df["unit_price"] <= 0

df.loc[invalid_price, "unit_price"] = df.loc[
    invalid_price, "base_unit_price"
]

# 2. Mark negative quantities as missing
df.loc[df["quantity"] < 0, "quantity"] = np.nan

# 3. Mark invalid expenses as missing
invalid_expenses = (
    (df["expenses"] < 0) |
    (df["expenses"] >= 99999999)
)

df.loc[invalid_expenses, "expenses"] = np.nan

# 4. Recalculate Revenue
valid_revenue = (
    df["quantity"].notna() &
    df["unit_price"].notna() &
    df["discount"].notna()
)

df.loc[valid_revenue, "revenue"] = (
    df.loc[valid_revenue, "quantity"] *
    df.loc[valid_revenue, "unit_price"]
    - df.loc[valid_revenue, "discount"]
)

# 5. Recalculate Profit
valid_profit = (
    df["revenue"].notna() &
    df["expenses"].notna()
)

df.loc[valid_profit, "profit"] = (
    df.loc[valid_profit, "revenue"] -
    df.loc[valid_profit, "expenses"]
)

print("Numeric Correction Completed!")
print("Missing Quantity:", df["quantity"].isna().sum())
print("Missing Unit Price:", df["unit_price"].isna().sum())
print("Missing Expenses:", df["expenses"].isna().sum())
print("Missing Revenue:", df["revenue"].isna().sum())
print("Missing Profit:", df["profit"].isna().sum())
```

```python
# Recover missing values using existing financial values

# 1. Recover Quantity
mask = (
    df["quantity"].isna()
    & df["revenue"].notna()
    & df["discount"].notna()
    & df["unit_price"].gt(0)
)

calculated_qty = (
    (df.loc[mask, "revenue"] + df.loc[mask, "discount"])
    / df.loc[mask, "unit_price"]
)

valid_qty = calculated_qty.notna() & (
    calculated_qty % 1 == 0
)

df.loc[calculated_qty[valid_qty].index, "quantity"] = (
    calculated_qty[valid_qty].astype("Int64")
)

# 2. Recover Discount
mask = (
    df["discount"].isna()
    & df["quantity"].notna()
    & df["unit_price"].notna()
    & df["revenue"].notna()
)

df.loc[mask, "discount"] = (
    df.loc[mask, "quantity"] * df.loc[mask, "unit_price"]
    - df.loc[mask, "revenue"]
)

# 3. Recover Expenses
mask = (
    df["expenses"].isna()
    & df["revenue"].notna()
    & df["profit"].notna()
)

df.loc[mask, "expenses"] = (
    df.loc[mask, "revenue"] - df.loc[mask, "profit"]
)

# Check remaining missing values
print("Remaining Missing Values:")
print(df[["quantity", "discount", "expenses"]].isna().sum())
```
